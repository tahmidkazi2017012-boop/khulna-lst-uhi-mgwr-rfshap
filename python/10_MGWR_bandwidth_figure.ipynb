# MGWR Bandwidth-Comparison Figure
Per-predictor MGWR bandwidths, 1990–2025, on a log scale. Bars near the total sample
size (n = 14,419) indicate a **global** process; small bars indicate a **local** process.

Font: **Times New Roman**. Saves `Fig_MGWR_bandwidths.png` (300 dpi) and `.pdf` to the
notebook folder. Self-contained — just run the cells top to bottom.

```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "Times New Roman"
plt.rcParams["mathtext.fontset"] = "stix"

# Per-predictor MGWR bandwidths (number of nearest neighbours), from the MGWR 2.2 summaries
data = {
    1990: {'NDVI': 375, 'NDBI': 466, 'SAVI': 3631, 'NDWI': 14417},
    1995: {'NDVI': 14417, 'NDBI': 333, 'SAVI': 14417, 'NDWI': 1176},
    2000: {'NDVI': 14417, 'NDBI': 1096, 'SAVI': 46, 'NDWI': 14417},
    2005: {'NDVI': 14417, 'NDBI': 454, 'SAVI': 14417, 'NDWI': 272},
    2010: {'NDVI': 14417, 'NDBI': 424, 'SAVI': 554, 'NDWI': 14417},
    2015: {'NDVI': 14417, 'NDBI': 345, 'SAVI': 159, 'NDWI': 14417},
    2020: {'NDVI': 14417, 'NDBI': 376, 'SAVI': 230, 'NDWI': 14417},
    2025: {'NDVI': 14417, 'NDBI': 915, 'SAVI': 276, 'NDWI': 14417},
}

years = [1990, 1995, 2000, 2005, 2010, 2015, 2020, 2025]
preds = ["NDBI", "SAVI", "NDVI", "NDWI"]     # local-ish -> global-ish
N = 14419
GLOBAL = 14417
colors = {"NDBI": "#d1495b", "SAVI": "#2a9d8f", "NDVI": "#457b9d", "NDWI": "#e9c46a"}
```

```python
fig, ax = plt.subplots(figsize=(11, 6.2))

x = np.arange(len(years))
nb = len(preds); w = 0.19

for i, p in enumerate(preds):
    vals = [data[y][p] for y in years]
    xpos = x + (i - (nb - 1) / 2) * w
    ax.bar(xpos, vals, width=w, color=colors[p], edgecolor="black",
           linewidth=0.6, label=p, zorder=3)
    for xi, v in zip(xpos, vals):
        if v >= GLOBAL - 5:
            ax.text(xi, v * 1.03, "Global", ha="center", va="bottom",
                    fontsize=8.5, rotation=90, zorder=5)
        else:
            ax.text(xi, v * 1.15, f"{int(v)}", ha="center", va="bottom",
                    fontsize=8.5, rotation=90, zorder=5)

ax.set_yscale("log")
ax.set_ylim(30, 60000)
ax.set_xlim(-0.6, len(years) - 0.5 + 0.55)
ax.set_xticks(x); ax.set_xticklabels([str(y) for y in years], fontsize=12)
ax.set_xlabel("Year", fontsize=13, labelpad=8)
ax.set_ylabel(f"MGWR bandwidth (nearest neighbours, log scale; global = n = {N:,})",
              fontsize=12, labelpad=8)
ax.set_title("Per-predictor MGWR bandwidths, 1990\u20132025\n"
             "(small = local process; near total sample size = global process)",
             fontsize=13.5, pad=12)

ax.axhline(GLOBAL, color="grey", ls="--", lw=1, zorder=1)
ax.text(len(years) - 0.5 + 0.62, GLOBAL, "Global\nscale",
        ha="left", va="center", fontsize=8.5, color="grey", style="italic")

ax.tick_params(axis="y", labelsize=11)
ax.grid(axis="y", which="major", ls=":", lw=0.5, color="grey", alpha=0.6, zorder=0)
ax.set_axisbelow(True)

ax.legend(title="Predictor", ncol=4, fontsize=11, title_fontsize=11.5,
          loc="upper center", bbox_to_anchor=(0.5, -0.11), frameon=False)

plt.subplots_adjust(bottom=0.20, top=0.86, left=0.09, right=0.93)
plt.savefig("Fig_MGWR_bandwidths.png", dpi=300)
plt.savefig("Fig_MGWR_bandwidths.pdf")
plt.show()
print("Saved Fig_MGWR_bandwidths.png and .pdf")
```
